# composer-nb examples

Run all cells with ⏩.

- `tempo 90` — a setting: a reserved word and its value. No `=`, because it isn't a name.
- `steps pair = X-x-`, `chords verse = Am F|C G`, `pattern groove = { … }` — a name, with its type in front. There are three types: steps, chords and patterns.
- `kick: x--x---` — inside braces, an instrument and what it plays. A drum takes steps, and `chords:` takes chords.
- `play { … }` — plays what's after it, a pattern or one line. Nothing plays unless it's inside a play.
- `loop` is the one way to repeat, and a length like `3 bars` is the one way to say how long.

Each play gives a sound you can play, save as a WAV, and see on a drum grid.

## What Happens Now?

A 7-step kick and snare figure loops against a 3-step ride bell. `groove` is 3 bars of that, which is where the loops line up. The play is 6 bars: one crash, with the groove looping under it.

```python
%%music
// what happens now: crash once, the groove looping under it
time 7 over 8

pattern groove = 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play 6 bars {
  crash: 1
  loop groove
}
```

A pattern and one line are the same thing, so a play with one thing in it needs no braces.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play groove
```

A pattern can play other patterns next to its own lines.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

pattern breakdown = 6 bars {
  crash: 1
  loop groove
}

play breakdown
```

Two plays, two outputs: the groove on its own, then with the crash.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play groove

play 6 bars {
  crash: 1
  loop groove
}
```

Two drums on the same step is two lines with a hit in the same place. Here the snare joins the kick on step 4.

```python
%%music
time 7 over 8

play 3 bars {
  kick: loop x--x---
  snare: loop ---xx--
}
```

## Names and their types

A name has its type in front: `steps` for a drum's hits, `chords`, or `pattern`. A name only goes where its type goes: steps on a drum's line, chords on the `chords:` line, and a pattern on a line of its own.

```python
%%music
sound epiano
steps pair = X-x-
chords verse = Am F|C G

pattern beat = {
  kick: x-------
  snare: ----x---
}

play {
  chords: verse
  hat: loop pair
  loop beat
}
```

## Lengths and loops

A length goes in front of what it measures, where you name a pattern or where you play it. `loop` repeats something until what it's in ends. Every number is arithmetic, so `*` and `/` only ever multiply and divide. That's why a time is written `7 over 8`: it's two numbers, not a fraction.

```python
%%music
time 7 over 8

pattern groove = {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play 3 bars groove
play (1+2)*2 bars groove
```

A loop has to fit what it's in a whole number of times. These loops line up every 3 bars, so 4 bars is a mistake, and the message says so.

```python
%%music
time 7 over 8

play 4 bars {
  kick: loop x--x---
  ride.bell: loop x--
}
```

Without a length, a pattern is as long as what's in it, down to one beat. Names in a row play one after another.

```python
%%music
pattern beat = {
  kick: x-------x-x-----
  snare: ----x-------x---
}
pattern fill = { snare: xxxx }
play beat fill beat
```

Brackets group, and a length with `loop` repeats what's after it: three bars of the kick figure, then a bar of crash and snare, and all of that twice.

```python
%%music
time 7 over 8
pattern fig = 1 bar { kick: loop x--x--- }
pattern hit = {
  crash: 1
  snare: 2 4
}
play 8 bars loop (3 bars loop fig hit)
```

## Chords

```python
%%music trains
// trains chorus
tempo 90
sound guitar
capo 5
play chords: A Am|C Cmaj7|D A|C Cmaj7|Cadd9|Em
```

`after trains` starts from the settings `trains` ended with, so this is still guitar. `intro` is a name for chords.

```python
%%music hotel after trains
// hotel california intro
tempo 75
capo 2
chords intro = Am E7|G D|F C|Dm E7
play chords: intro intro
```

## Chords and drums together

```python
%%music
sound epiano
play {
  chords: Am F|C G
  kick: loop x-------
  snare: loop ----x---
  hat: loop x-
}
```

## From Python

Songs are Python values, so you can build them with code.

```python
from composer_nb import Song

blues = "|".join(["A7", "D7", "A7", "A7", "D7", "D7", "A7", "A7", "E7", "D7", "A7", "E7"])
Song("tempo 120\nsound epiano\nplay chords: " + blues)
```
