# composer-nb examples

Run all cells, then press play on any of them. `%%music name` saves a cell under a name, and `%%music b after a` starts `b` from the settings `a` ended with.

```python
%load_ext composer_nb
```

## Drums

What Happens Now? (Porcupine Tree): a 7-step kick and snare against a 3-step ride bell, in 7/8.

```python
%%music
-- what happens now
time: 7/8
crash: 1|.|.
ride: b..
kick: x..x...
snare: ....x..
```

The same groove twice. `repeat 2:` covers the lines indented under it.

```python
%%music
time: 7/8
repeat 2:
  crash: 1|.|.
  ride: b..
  kick: x..x...
  snare: ....x..
```

Repeats nest: the kick figure three times, then a bar of crash and snare, all twice, then one bar of snare.

```python
%%music
time: 7/8
repeat 2:
  repeat 3:
    kick: x..x...
  crash: 1
  snare: 2 4
snare: 1 2 3 4
```

## Chords

```python
%%music trains
-- trains chorus
tempo: 90
capo: 5
sound: guitar
A Am|C Cmaj7|D A|C Cmaj7|Cadd9|Em
```

`after trains` keeps the guitar. This cell sets its own tempo and capo.

```python
%%music hotel after trains
-- hotel california intro
tempo: 75
capo: 2
repeat 2:
  Am E7|G D|F C|Dm E7
```

## Chords and drums together

```python
%%music
tempo: 100
sound: epiano
Am F|C G
kick: 1 3
snare: 2 4
hat: x.
```

## From Python

Songs are Python values, so you can build them with code.

```python
from composer_nb import Song

blues = "|".join(["A7", "D7", "A7", "A7", "D7", "D7", "A7", "A7", "E7", "D7", "A7", "E7"])
Song("tempo: 120\nsound: epiano\n" + blues)
```
