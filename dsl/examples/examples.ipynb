# composer-nb examples

Run all cells with ⏩.

- `tempo 90` — a setting: a reserved word and its value. No `=`, because it isn't a variable.
- `verse = Am F|C G` — a variable: your own name for steps or chords. Only these get `=`.
- `pattern groove { … }` — a pattern: lines that play together. Braces always follow a keyword.
- `kick: x--x---` — inside braces, an instrument and what it plays. One instrument per line.
- `play { … }` — plays what's after it, a block or one line. Nothing plays unless it's inside a play.

Each play gives a sound you can play, save as a WAV, and see on a drum grid.

```python
%load_ext composer_nb
```

## What Happens Now?

A 7-step kick and snare figure loops against a 3-step ride bell. `groove` is 3 bars of that. The play block plays the crash once and the groove twice, together.

```python
%%music
// what happens now: crash once, groove twice
time 7 over 8

pattern groove 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play {
  crash: 1
  groove * 2
}
```

A block and one line are the same thing, so a play with one thing in it needs no braces.

```python
%%music
time 7 over 8

pattern groove 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play groove
```

A pattern can play other patterns next to its own lines.

```python
%%music
time 7 over 8

pattern groove 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

pattern breakdown {
  crash: 1
  groove * 2
}

play breakdown
```

Two plays, two outputs: the groove on its own, then with the crash.

```python
%%music
time 7 over 8

pattern groove 3 bars {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play groove

play {
  crash: 1
  groove * 2
}
```

Two drums on the same step is two lines with a hit in the same place. Here the snare joins the kick on step 4.

```python
%%music
time 7 over 8

play 3 bars {
  kick: loop x--x---
  snare: loop ---xx--
}
```

## Lengths and numbers

A length goes in front of what it measures, where you define a pattern or where you play it. Every number is arithmetic, so `/` only ever divides. That's why a time is written `7 over 8`: it's two numbers, not a fraction.

```python
%%music
time 7 over 8

pattern groove {
  kick: loop x--x---
  snare: loop ----x--
  ride.bell: loop x--
}

play 3 bars groove
play (1+2) bars groove * (4/2)
```

Without a length, a pattern is as long as what's in it, down to one beat.

```python
%%music
pattern beat {
  kick: x-------x-x-----
  snare: ----x-------x---
}
pattern fill { snare: xxxx }
play beat fill beat
```

Brackets group: the kick figure three times, then a bar of crash and snare, all of that twice.

```python
%%music
time 7 over 8
pattern fig 1 bar { kick: loop x--x--- }
pattern hit {
  crash: 1
  snare: 2 4
}
play (fig * 3 hit) * 2
```

## Chords

```python
%%music trains
// trains chorus
tempo 90
sound guitar
capo 5
play chords: A Am|C Cmaj7|D A|C Cmaj7|Cadd9|Em
```

`after trains` starts from the settings `trains` ended with, so this is still guitar. `intro` is a variable holding chords.

```python
%%music hotel after trains
// hotel california intro
tempo 75
capo 2
intro = Am E7|G D|F C|Dm E7
play chords: intro intro
```

## Chords and drums together

```python
%%music
sound epiano
play {
  chords: Am F|C G
  kick: loop x-------
  snare: loop ----x---
  hat: loop x-
}
```

## From Python

Songs are Python values, so you can build them with code.

```python
from composer_nb import Song

blues = "|".join(["A7", "D7", "A7", "A7", "D7", "D7", "A7", "A7", "E7", "D7", "A7", "E7"])
Song("tempo 120\nsound epiano\nplay chords: " + blues)
```
