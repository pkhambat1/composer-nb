# composer-nb examples

Run all cells with ⏩.

- `tempo 90` — a setting: a reserved word and its value. No `=`, because it isn't a name.
- `steps pair = ^-x-`, `notes riff = E---F# E-E--`, `chords verse = Am F|C G`, `pattern groove = { … }` — a name, with its type in front. There are four types: steps, notes, chords and patterns.
- `kick: x--x---`, `guitar: E---F# E-E--`, `piano: chords Am F|C G` — inside braces, an instrument and what it plays. A drum takes steps. A pitched instrument takes notes, drawn on the same steps, or `chords` and a chart.
- `play { … }` — plays what's after it, a pattern or one line. Nothing plays unless it's inside a play.
- A pattern repeats everything in it until it ends, and a play's own lines play once. A length like `3 bars` is the one way to say how long, and in front of a pattern it repeats the pattern for that long.

Each play gives a sound you can play, save as a WAV, and see on a grid.

## What Happens Now?

A 7-step kick and snare figure repeats against a 3-step ride bell. `groove` is 3 bars of that, which is where they line up. The play is 6 bars: one crash, with the groove repeating under it.

```python
%%music
// what happens now: crash once, the groove repeating under it
time 7 over 8

pattern groove = 3 bars {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
}

play 6 bars {
  crash: 1
  groove
}
```

A pattern and one line are the same thing, so a play with one thing in it needs no braces.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
}

play groove
```

A pattern can play other patterns next to its own lines. `bar 1 crash: 1` puts the crash in the pattern's first bar only, so it comes back each time the pattern does.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
}

pattern breakdown = {
  bar 1 crash: 1
  groove
}

play 6 bars breakdown
```

Two plays, two outputs: the groove on its own, then with the crash.

```python
%%music
time 7 over 8

pattern groove = 3 bars {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
}

play groove

play 6 bars {
  crash: 1
  groove
}
```

Two drums on the same step is two lines with a hit in the same place. Here the snare joins the kick on step 4.

```python
%%music
time 7 over 8

pattern hits = {
  kick: x--x---
  snare: ---xx--
}

play 3 bars hits
```

## One bar only

Every line in a pattern repeats, so it plays in every bar. `bar 2 { ... }` is what plays in bar 2 only: here a burst of ghost notes, an open hi-hat, and the foot closing it.

Inside it, a length in front of `rest` stands for a run of rests, and commas separate the parts of a line: `2 beats rest, dd, 1 beats rest` is two beats of rest, the doubles, and one beat of rest. Lengths are always plural, whatever the number.

```python
%%music
time 7 over 8
tempo 90

pattern groove = {
  kick: x--x---
  snare: 2e 4
  snare.ghost: --x----
  hat: ^-x-
  bar 2 {
    snare.ghost: 2 beats rest, dd, 1 beats rest
    hat.open: 2 beats rest, --^-, 2 steps rest
    hat.pedal: 4
  }
}

play 6 bars {
  crash: 1
  groove
}
```

## Names and their types

A name has its type in front: `steps` for a drum's hits, `notes`, `chords`, or `pattern`. A name only goes where its type goes: steps on a drum's line, notes on a pitched instrument's line, chords after `chords` on one, and a pattern on a line of its own.

```python
%%music
steps pair = ^-x-
notes line = A1------F1------C2------G1------
chords verse = Am F|C G

pattern beat = {
  kick: x-------
  snare: ----x---
  hat: pair
}

play {
  piano.electric: chords verse
  bass: line
  beat
}
```

## Lengths and repeats

A length goes in front of what it measures, where you name a pattern or where you play it. Everything in a pattern repeats until the pattern ends, and a length in front of a pattern repeats it for that long. Every number is arithmetic, so `*` and `/` only ever multiply and divide. That's why a time is written `7 over 8`: it's two numbers, not a fraction.

```python
%%music
time 7 over 8

pattern groove = {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
}

play 3 bars groove
play (1+2)*2 bars groove
```

Whatever repeats has to fit what it's in a whole number of times. These lines line up every 3 bars, so 4 bars is a mistake, and the message says so.

```python
%%music
time 7 over 8

pattern groove = 4 bars {
  kick: x--x---
  ride.bell: x--
}

play groove
```

Without a length, a pattern lasts until its lines line up again. Names in a row play one after another.

```python
%%music
pattern beat = {
  kick: x-------x-x-----
  snare: ----x-------x---
}
pattern fill = { snare: xxxx }
play beat fill beat
```

Brackets group, and a length in front repeats what's after it: three bars of the kick figure, then a bar of crash and snare, and all of that twice.

```python
%%music
time 7 over 8
pattern fig = 1 bars { kick: x--x--- }
pattern hit = {
  crash: 1
  snare: 2 4
}
play 8 bars (3 bars fig hit)
```

## Notes

A pitched instrument's line is drawn on the same steps as a drum's, with a note where the drum has its `x`. A note is a capital letter, with `#` or `b` and an octave when it needs one. `-` lets it ring, `_` stops it, and a space goes between two notes that would otherwise touch (`F# E`).

This riff is 10 steps against the drums' 7 and 3. With no lengths written, the pattern lasts until all four lines meet, which is 15 bars. `octave 2` puts it on the guitar's low E string.

```python
%%music
time 7 over 8
octave 2

pattern groove = {
  kick: x--x---
  snare: ----x--
  ride.bell: x--
  guitar.electric: E---F# E-E--
}

play {
  crash: 1
  groove
}
```

A line plays one note at a time. For more, the instrument's line takes a block of lines, which play together.

```python
%%music
play piano: {
  C5-D5-E5-G5-E5-D5-C5---
  C3---------------
}
```

## Chords

Chords are written by the bar, on their instrument's line with `chords` in front. A capo moves a guitar's chords up the neck.

```python
%%music trains
// trains chorus
tempo 90
capo 5
play guitar: chords A Am|C Cmaj7|D A|C Cmaj7|Cadd9|Em
```

`after trains` starts from the settings `trains` ended with, and changes the ones it sets again. `intro` is a name for chords.

```python
%%music hotel after trains
// hotel california intro
tempo 75
capo 2
chords intro = Am E7|G D|F C|Dm E7
play guitar: chords intro intro
```
